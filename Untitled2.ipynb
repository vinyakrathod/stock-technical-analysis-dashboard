```python

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import yfinance as yf
import plotly.graph_objects as go
from IPython.display import display

# ============================================================
# 1. USER INPUTS
# ============================================================

ticker = input("Enter company ticker (e.g. RELIANCE.NS, TCS.NS, AAPL): ").strip().upper()
timeframe = input("Enter timeframe (1d, 1wk, 1mo, 1h, 30m, 15m, 5m): ").strip().lower()
duration = input("Enter duration (1mo, 3mo, 6mo, 1y, 2y, 5y, max): ").strip().lower()

if not ticker:
    raise ValueError("Please enter a valid ticker.")

INTERVALS = {
    "1m": ("1m", "7d", "5m"),
    "2m": ("2m", "60d", "5m"),
    "5m": ("5m", "60d", "15m"),
    "15m": ("15m", "60d", "1h"),
    "30m": ("30m", "60d", "1h"),
    "60m": ("60m", "60d", "1d"),
    "1h": ("60m", "60d", "1d"),
    "90m": ("90m", "60d", "1d"),
    "1d": ("1d", "max", "1wk"),
    "5d": ("5d", "max", "1mo"),
    "1wk": ("1wk", "max", "1mo"),
    "1mo": ("1mo", "max", "3mo"),
}

DURATIONS = {
    "1d": "1d", "5d": "5d", "1mo": "1mo",
    "3mo": "3mo", "6mo": "6mo", "1y": "1y",
    "2y": "2y", "5y": "5y", "10y": "10y",
    "ytd": "ytd", "max": "max"
}

if timeframe not in INTERVALS:
    raise ValueError("Unsupported timeframe. Choose: " + ", ".join(INTERVALS))

if duration not in DURATIONS:
    raise ValueError("Unsupported duration. Choose: " + ", ".join(DURATIONS))

interval, max_period, htf_interval = INTERVALS[timeframe]
period = DURATIONS[duration]

# Yahoo Finance limits intraday historical data.
if interval in {"1m", "2m", "5m", "15m", "30m", "60m", "90m"}:
    allowed = {
        "1m": "7d",
        "2m": "60d",
        "5m": "60d",
        "15m": "60d",
        "30m": "60d",
        "60m": "730d",
        "90m": "60d"
    }
    if period == "max" or period in {
        "1mo", "3mo", "6mo", "1y", "2y", "5y", "10y", "ytd"
    }:
        print("Note: Intraday history is restricted by Yahoo Finance.")
        period = allowed[interval]

print(f"\nDownloading {ticker} | Interval: {interval} | Period: {period}")

# ============================================================
# 2. DOWNLOAD STOCK DATA
# ============================================================

df = yf.download(
    ticker,
    period=period,
    interval=interval,
    auto_adjust=True,
    progress=False,
    threads=False
)

if df.empty:
    raise ValueError(
        "No data received. Check the ticker, internet connection, "
        "and Yahoo Finance history limits."
    )

# Handle both ordinary and MultiIndex columns.
if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df.columns = [str(c).title() for c in df.columns]
df = df.loc[:, ~df.columns.duplicated()]

required = ["Open", "High", "Low", "Close"]
missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing required stock columns: {missing}")

df = df.dropna(subset=required).sort_index()
df = df[~df.index.duplicated(keep="last")].copy()

if len(df) < 2:
    raise ValueError("Not enough price candles to create the charts.")

# Convert price columns to numeric.
for col in required:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=required)

if df.empty:
    raise ValueError("No valid price candles remain after cleaning.")

# ============================================================
# 3. CALCULATE TECHNICAL INDICATORS
# ============================================================

close = df["Close"]
high = df["High"]
low = df["Low"]

df["SMA50"] = close.rolling(50, min_periods=50).mean()
df["EMA20"] = close.ewm(span=20, adjust=False, min_periods=20).mean()
df["EMA50"] = close.ewm(span=50, adjust=False, min_periods=50).mean()

previous_close = close.shift(1)

true_range = pd.concat(
    [
        high - low,
        (high - previous_close).abs(),
        (low - previous_close).abs()
    ],
    axis=1
).max(axis=1)

df["ATR14"] = true_range.rolling(14, min_periods=14).mean()

# Stochastic oscillator: %K(20), %D(3).
lowest20 = low.rolling(20, min_periods=20).min()
highest20 = high.rolling(20, min_periods=20).max()
denominator = (highest20 - lowest20).replace(0, np.nan)

df["Stoch_K20"] = 100 * (close - lowest20) / denominator
df["Stoch_D3"] = df["Stoch_K20"].rolling(3, min_periods=3).mean()

# RSI(14), using Wilder-style exponential smoothing.
delta = close.diff()
gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)

avg_gain = gain.ewm(
    alpha=1/14, adjust=False, min_periods=14
).mean()

avg_loss = loss.ewm(
    alpha=1/14, adjust=False, min_periods=14
).mean()

rs = avg_gain / avg_loss.replace(0, np.nan)
df["RSI14"] = 100 - (100 / (1 + rs))

df.loc[(avg_loss == 0) & (avg_gain > 0), "RSI14"] = 100
df.loc[(avg_loss == 0) & (avg_gain == 0), "RSI14"] = 50

# ============================================================
# 4. FIND DEMAND AND SUPPLY ZONES
# ============================================================

def find_zones(data, max_base_candles=3):
    zones = []
    d = data.reset_index()

    for base_start in range(3, len(d) - 1):
        for n_base in range(1, max_base_candles + 1):

            base_end = base_start + n_base - 1
            leg_idx = base_end + 1

            if leg_idx >= len(d):
                continue

            atr = data["ATR14"].iloc[leg_idx]

            if pd.isna(atr) or atr <= 0:
                continue

            base = d.iloc[base_start:base_end + 1]
            leg = d.iloc[leg_idx]

            base_high = float(base["High"].max())
            base_low = float(base["Low"].min())

            # Base must be relatively small compared with ATR.
            if base_high - base_low > 1.5 * atr:
                continue

            body = abs(float(leg["Close"] - leg["Open"]))

            # Confirm a strong leg-out.
            bearish_break = float(leg["Close"]) < base_low
            bullish_break = float(leg["Close"]) > base_high
            strong_body = body >= 1.2 * atr

            # Permit a directional sequence of 2-3 candles.
            seq_end = min(leg_idx + 2, len(d) - 1)
            seq = d.iloc[leg_idx:seq_end + 1]

            bearish_seq = (
                len(seq) >= 2
                and (seq["Close"] < seq["Open"]).all()
            )

            bullish_seq = (
                len(seq) >= 2
                and (seq["Close"] > seq["Open"]).all()
            )

            bearish = bearish_break and (strong_body or bearish_seq)
            bullish = bullish_break and (strong_body or bullish_seq)

            # Classify the approach into the base.
            prior = d.iloc[max(0, base_start - 3):base_start]

            if len(prior) >= 2:
                prior_move = float(
                    prior["Close"].iloc[-1] - prior["Close"].iloc[0]
                )
            else:
                prior_move = 0.0

            date_value = d.iloc[base_end, 0]

            if bearish:
                lower = float(
                    np.minimum(base["Open"], base["Close"]).min()
                )
                upper = base_high

                if lower < upper:
                    pattern = "RBD" if prior_move >= 0 else "DBD"
                    strength = body / atr + max(
                        0, (base_low - float(leg["Close"])) / atr
                    )

                    zones.append({
                        "kind": "Supply",
                        "lower": lower,
                        "upper": upper,
                        "formed_at": date_value,
                        "strength": float(strength),
                        "pattern": pattern,
                        "base_end": base_end,
                        "leg_idx": leg_idx
                    })

            if bullish:
                lower = base_low
                upper = float(
                    np.maximum(base["Open"], base["Close"]).max()
                )

                if lower < upper:
                    pattern = "RBR" if prior_move >= 0 else "DBR"
                    strength = body / atr + max(
                        0, (float(leg["Close"]) - base_high) / atr
                    )

                    zones.append({
                        "kind": "Demand",
                        "lower": lower,
                        "upper": upper,
                        "formed_at": date_value,
                        "strength": float(strength),
                        "pattern": pattern,
                        "base_end": base_end,
                        "leg_idx": leg_idx
                    })

    # Keep the strongest instance of near-identical zones.
    zones.sort(key=lambda z: z["strength"], reverse=True)
    deduped = []

    for zone in zones:
        width = max(zone["upper"] - zone["lower"], 1e-9)

        duplicate = any(
            existing["kind"] == zone["kind"]
            and abs(existing["lower"] - zone["lower"]) <= width * 0.05
            and abs(existing["upper"] - zone["upper"]) <= width * 0.05
            for existing in deduped
        )

        if not duplicate:
            deduped.append(zone)

    # A zone is fresh when no subsequent candle touches its range.
    for zone in deduped:
        later = data.iloc[zone["leg_idx"] + 1:]

        if later.empty:
            zone["fresh"] = True
        else:
            overlap = (
                (later["Low"] <= zone["upper"])
                & (later["High"] >= zone["lower"])
            )
            zone["fresh"] = not bool(overlap.any())

    return deduped


zones = find_zones(df)
cmp = float(df["Close"].iloc[-1])

# Supply must be above CMP; demand must be below CMP.
supplies = [
    z for z in zones
    if z["kind"] == "Supply"
    and z["fresh"]
    and z["lower"] > cmp
]

demands = [
    z for z in zones
    if z["kind"] == "Demand"
    and z["fresh"]
    and z["upper"] < cmp
]

supplies.sort(
    key=lambda z: (-z["strength"], -pd.Timestamp(z["formed_at"]).timestamp())
)

demands.sort(
    key=lambda z: (-z["strength"], -pd.Timestamp(z["formed_at"]).timestamp())
)

# ============================================================
# 5. DISPLAY ZONE RESULTS
# ============================================================

print("\n" + "=" * 60)
print(f"STOCK: {ticker}")
print(f"TIMEFRAME: {timeframe}")
print(f"DURATION: {duration}")
print(f"CURRENT MARKET PRICE (latest close): {cmp:.2f}")
print("=" * 60)

def show_zones(title, selected_zones):
    print(f"\n{title}")

    if not selected_zones:
        print("No qualifying fresh zones found.")
        return

    for z in selected_zones[:10]:
        print(
            f'{z["lower"]:.2f} - {z["upper"]:.2f} | '
            f'{z["pattern"]} | Strength: {z["strength"]:.2f} | '
            f'Formed: {z["formed_at"]}'
        )

show_zones("FRESH SUPPLY ZONES ABOVE CMP", supplies)
show_zones("FRESH DEMAND ZONES BELOW CMP", demands)

# ============================================================
# 6. CHART 1: CANDLESTICKS + DEMAND/SUPPLY ZONES
# ============================================================

fig1 = go.Figure()

fig1.add_trace(go.Candlestick(
    x=df.index,
    open=df["Open"],
    high=df["High"],
    low=df["Low"],
    close=df["Close"],
    name="Price"
))

for z in supplies + demands:
    is_supply = z["kind"] == "Supply"

    fill = (
        "rgba(220,53,69,0.15)"
        if is_supply else "rgba(25,135,84,0.15)"
    )

    line = (
        "rgba(220,53,69,0.85)"
        if is_supply else "rgba(25,135,84,0.85)"
    )

    fig1.add_hrect(
        y0=z["lower"],
        y1=z["upper"],
        fillcolor=fill,
        line_color=line,
        line_width=1,
        annotation_text=(
            f'{z["kind"]} {z["pattern"]}: '
            f'{z["lower"]:.2f}-{z["upper"]:.2f}'
        ),
        annotation_position="top left"
    )

fig1.update_layout(
    title=f"{ticker}: Price with Fresh Demand/Supply Zones ({timeframe})",
    xaxis_title="Date",
    yaxis_title="Price",
    xaxis_rangeslider_visible=False,
    template="plotly_white",
    hovermode="x unified",
    height=650
)

fig1.show()

# ============================================================
# 7. CHART 2: SMA 50 + EMA 20 + EMA 50
# ============================================================

fig2 = go.Figure()

fig2.add_trace(go.Scatter(
    x=df.index, y=df["Close"],
    name="Close Price", mode="lines"
))

fig2.add_trace(go.Scatter(
    x=df.index, y=df["SMA50"],
    name="SMA 50", mode="lines"
))

fig2.add_trace(go.Scatter(
    x=df.index, y=df["EMA20"],
    name="EMA 20", mode="lines"
))

fig2.add_trace(go.Scatter(
    x=df.index, y=df["EMA50"],
    name="EMA 50", mode="lines"
))

fig2.update_layout(
    title=f"{ticker}: Moving Averages ({timeframe})",
    xaxis_title="Date",
    yaxis_title="Price",
    template="plotly_white",
    hovermode="x unified",
    height=550
)

fig2.show()

# ============================================================
# 8. CHART 3: STOCHASTIC OSCILLATOR (20, 3)
# ============================================================

fig3 = go.Figure()

fig3.add_trace(go.Scatter(
    x=df.index, y=df["Stoch_K20"],
    name="%K (20)", mode="lines"
))

fig3.add_trace(go.Scatter(
    x=df.index, y=df["Stoch_D3"],
    name="%D (3)", mode="lines"
))

fig3.add_hline(
    y=80, line_dash="dash",
    annotation_text="Overbought (80)"
)

fig3.add_hline(
    y=20, line_dash="dash",
    annotation_text="Oversold (20)"
)

fig3.update_layout(
    title=f"{ticker}: Stochastic Oscillator (20, 3)",
    xaxis_title="Date",
    yaxis_title="Stochastic",
    yaxis=dict(range=[0, 100]),
    template="plotly_white",
    hovermode="x unified",
    height=450
)

fig3.show()

# ============================================================
# 9. CHART 4: RSI (14)
# ============================================================

fig4 = go.Figure()

fig4.add_trace(go.Scatter(
    x=df.index, y=df["RSI14"],
    name="RSI 14", mode="lines"
))

fig4.add_hline(
    y=70, line_dash="dash",
    annotation_text="Overbought (70)"
)

fig4.add_hline(
    y=30, line_dash="dash",
    annotation_text="Oversold (30)"
)

fig4.update_layout(
    title=f"{ticker}: Relative Strength Index (14)",
    xaxis_title="Date",
    yaxis_title="RSI",
    yaxis=dict(range=[0, 100]),
    template="plotly_white",
    hovermode="x unified",
    height=450
)

fig4.show()

# ============================================================
# 10. SUMMARY
# ============================================================

summary = pd.DataFrame({
    "Indicator": [
        "Latest Close",
        "SMA 50",
        "EMA 20",
        "EMA 50",
        "RSI 14",
        "Stochastic %K (20)",
        "Stochastic %D (3)",
        "ATR 14"
    ],
    "Latest Value": [
        df["Close"].iloc[-1],
        df["SMA50"].iloc[-1],
        df["EMA20"].iloc[-1],
        df["EMA50"].iloc[-1],
        df["RSI14"].iloc[-1],
        df["Stoch_K20"].iloc[-1],
        df["Stoch_D3"].iloc[-1],
        df["ATR14"].iloc[-1]
    ]
})

summary["Latest Value"] = summary["Latest Value"].round(2)

print("\nLATEST TECHNICAL INDICATOR VALUES")
display(summary)

print("\nDashboard completed.")
```
